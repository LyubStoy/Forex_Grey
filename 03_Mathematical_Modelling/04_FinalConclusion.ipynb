\quad \textbf{Final Conclusion — EUR/USD vs Yield Spread Analysis} \\[1.2em]
\quad \text{The empirical analysis of the relationship between changes in the EUR/USD exchange} \\
\quad \text{rate and changes in the US--Euro Area yield spread provides robust evidence of a} \\
\quad \textbf{statistically significant}, \text{ yet economically moderate, relationship.} \\[1.2em]
\quad \text{The Pearson correlation coefficient, computed on first differences, is negative and} \\
\quad \text{highly significant across the full sample and sub-periods, indicating that increases} \\
\quad \text{in the yield spread are systematically associated with declines in the EUR/USD} \\
\quad \text{exchange rate. This finding is consistent with standard macro-financial theory,} \\
\quad \text{whereby higher relative US yields attract capital inflows and strengthen the US dollar.} \\[1.2em]
\quad \text{To move beyond descriptive statistics, a linear regression framework was employed} \\
\quad \text{to quantify this relationship. The estimated coefficient associated with the yield} \\
\quad \text{spread is negative and statistically significant, confirming that changes in the} \\
\quad \text{spread exert a measurable downward pressure on EUR/USD. The confidence interval} \\
\quad \text{lies entirely in the negative region, indicating that the relationship is both} \\
\quad \textbf{stable and robust}. \\[1.2em]
\quad \text{The intercept term is not statistically different from zero, suggesting the absence} \\
\quad \text{of systematic drift in exchange rate changes when the yield spread remains constant.} \\
\quad \text{This supports the use of first differences and highlights that exchange rate dynamics} \\
\quad \text{are primarily driven by shocks rather than deterministic trends.} \\[1.2em]
\quad \text{The overall regression model is globally significant, as indicated by a large F-statistic} \\
\quad \text{and an associated p-value effectively equal to zero. This confirms that the explanatory} \\
\quad \text{variable contains genuine information and that the observed relationship is not due} \\
\quad \text{to random variation.} \\[1.2em]
\quad \text{Residual diagnostics further validate the model specification. The Durbin--Watson} \\
\quad \text{statistic is close to two, indicating no evidence of autocorrelation and suggesting} \\
\quad \text{that residuals are temporally independent. However, the normality assumption is} \\
\quad \text{rejected, with residuals exhibiting excess kurtosis and fat tails. This reflects} \\
\quad \text{the presence of extreme movements and is consistent with well-established stylized} \\
\quad \text{facts of financial time series.} \\[1.2em]
\quad \textbf{Despite the statistical significance of the results, the magnitude of the relationship} \\
\quad \textbf{remains relatively modest, implying that the yield spread explains only a limited} \\
\quad \textbf{portion of the overall variability in EUR/USD. This highlights the importance of} \\
\quad \textbf{additional macroeconomic and financial drivers, such as inflation differentials,} \\
\quad \textbf{monetary policy expectations, and geopolitical shocks.}
