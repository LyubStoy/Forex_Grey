\qquad\quad\quad\Large\textbf{Exchange Rate Construction}\\[2em]
\qquad\quad\quad\textbf{A.}\qquad \textbf{Economy reasons}\\[1em]
\qquad\quad\quad \text{The most common explanation for the formation of the exchange rate EUR/USD is the difference between}\\
\qquad\quad\quad \text{the monetary policy of the FED (the Federal Reserve of US – it is like a central bank in the USA)}\\
\qquad\quad\quad \text{and the interest rate in the EU decided by the ECB (European Central Bank).}\\
\qquad\quad\quad \text{These two entities are the final decision makers and their decisions are based on the real economy.}\\
\qquad\quad\quad \text{ So, the relation must be explained in this way:}\\[1em]
\qquad\quad\quad\quad \mathbf{Economy \;\rightarrow\; Inflation \;\rightarrow\; Central\ Bank \;\rightarrow\; Expectations \;\rightarrow\; Forex}\\[1em]
\qquad\quad\quad \text{Therefore, the core of the object is the real economy. The market is following the real economy via the periodical economic }\\
\qquad\quad\quad \text{indicators and tries to anticipate the future moves of the interest rate decided by the respective Central Bank.}\\
\qquad\quad\quad \text{The most important indicators are the inflation ones:}\\[1em]
\qquad\quad\quad\quad\quad \underline{\mathbf{INFLATION\ INDICATORS\ IN\ US\ AND\ EUROPE}}\\[2em]
\qquad\qquad\quad\quad\quad\quad \textbf{US:}\\
\qquad\quad\quad\quad\quad\quad\quad\quad\quad\quad \bullet \quad \text{CPI (Consumer Price Index)}\\
\qquad\quad\quad\quad\quad\quad\quad\quad\quad\quad \bullet \quad \text{Core CPI}\\
\qquad\quad\quad\quad\quad\quad\quad\quad\quad\quad \bullet \quad \text{PCE (Personal Consumption Expenditures)}\\[1em]
\qquad\qquad\quad\quad\quad\quad \textbf{Europe:}\\
\qquad\quad\quad\quad\quad\quad\quad\quad\quad\quad\quad\quad \bullet \quad \text{HICP (inflation eurozone)}\\[1em]
\qquad\quad\quad\quad\quad\quad\quad\quad \text{The logic here is if inflation is up there is expectation for rate hikes and the currency}\\
\qquad\quad\quad\quad\quad\quad\quad\quad \text{gets more attractive = more expensive.}\\[3em]
\qquad\quad\quad \text{Eventually, one can state that expectations for future rate decisions and movements determine}\\
\qquad\quad\quad \text{the exchange rate. More precisely, the pricing of the future interest rate differential produces}\\
\qquad\quad\quad \text{the exchange rate between the two currencies, in this case: EUR/USD.}\\[1em]
\qquad\quad\quad \textbf{To put it simply: it can be stated that the anticipated difference between the interest rates}\\
\qquad\quad\quad \textbf{of two countries determines the Forex (Foreign Exchange Rate) of their currencies.}\\[2em]
\qquad\quad\quad \textbf{In Europe and in US, this anticipation can be represented and followed by:}\\[1em]
\qquad\quad\quad 1.\quad \textbf{Government Bond Yields}\\[1em]
\qquad\quad\quad\quad \circ \quad \textbf{German Bond Yield}\\[0.5em]
\qquad\quad\quad\quad\quad \bullet \quad \text{German 2-year yield (VERY important)}\\
\qquad\quad\quad\quad\quad \bullet \quad \text{German 10-year yield}\\[1em]
\qquad\quad\quad\quad \circ \quad \textbf{US Treasury Yield}\\[0.5em]
\qquad\quad\quad\quad\quad \bullet \quad \text{US 2-year yield (VERY important)}\\
\qquad\quad\quad\quad\quad \bullet \quad \text{US 10-year yield}\\[1.5em]
\qquad\quad\quad 2.\quad \textbf{Yield Curve (shape matters)}\\[1em]
\qquad\quad\quad\quad\quad \textbf{Not just levels — the curve structure}\\[1em]
\qquad\quad\quad\quad\quad\quad \bullet \quad \text{inverted } \rightarrow \text{ recession + future rate cuts expected}\\
\qquad\quad\quad\quad\quad\quad \bullet \quad \text{steep } \rightarrow \text{ growth + rate hikes expected}\\[1.5em]
\qquad\quad\quad \textbf{In conclusion, these two comparisons explain a large part of EUR/USD moves:}\\[1em]
