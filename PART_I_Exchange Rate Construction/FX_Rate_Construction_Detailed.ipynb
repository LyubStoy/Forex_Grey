\qquad\quad\quad\Large\textbf{Exchange Rate Construction}\\[2em]
\qquad\quad\quad \textbf{A.}\qquad \textbf{Economy reasons}\\[1em]
\qquad\quad\quad \text{The most common explanation for the formation of the exchange rate EUR/USD is the difference}\\
\qquad\quad\quad \text{between the monetary policy of the FED (Federal Reserve of the US) and the interest rate in}\\
\qquad\quad\quad \text{the EU decided by the ECB (European Central Bank). These two entities are the final decision}\\
\qquad\quad\quad \text{makers and their decisions are based on the real economy.}\\[1em]
\qquad\quad\quad \text{So, the relation can be expressed as:}\\[1em]
\qquad\quad\quad\quad \mathbf{Economy \rightarrow Inflation \rightarrow Central\ Bank \rightarrow Expectations \rightarrow Forex}\\[1em]
\qquad\quad\quad \text{Therefore, the core of the system is the real economy. The market follows economic indicators}\\
\qquad\quad\quad \text{and anticipates future interest rate movements. These indicators can be grouped as follows:}\\[2em]
%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
\qquad\quad\quad \underline{\mathbf{INFLATION\ INDICATORS\ (US\ \&\ EUROPE)}}\\[1em]
\qquad\quad\quad \textbf{US:}\\
\qquad\quad\quad\quad \bullet \quad \text{CPI (Consumer Price Index)}\\
\qquad\quad\quad\quad \bullet \quad \text{Core CPI}\\
\qquad\quad\quad\quad \bullet \quad \text{PCE (VERY important for the Fed)}\\[1em]
\qquad\quad\quad \textbf{Europe:}\\
\qquad\quad\quad\quad \bullet \quad \text{HICP (Eurozone inflation)}\\[1em]
\qquad\quad\quad \text{Logic: inflation up } \rightarrow \text{ rate hikes expected } \rightarrow \text{ currency stronger}\\[2em]
%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
\qquad\quad\quad \underline{\mathbf{LABOUR\ MARKET\ INDICATORS}}\\[1em]
\qquad\quad\quad \text{Especially important for the US}\\[1em]
\qquad\quad\quad \textbf{US:}\\
\qquad\quad\quad\quad \bullet \quad \text{NFP (Non-Farm Payrolls)}\\
\qquad\quad\quad\quad \bullet \quad \text{Unemployment rate}\\
\qquad\quad\quad\quad \bullet \quad \text{Average hourly earnings (wages)}\\[1em]
\qquad\quad\quad \textbf{Europe:}\\
\qquad\quad\quad\quad \bullet \quad \text{Unemployment rate (less reactive)}\\[1em]
\qquad\quad\quad \text{Logic: strong jobs } \rightarrow \text{ strong economy } \rightarrow \text{ inflation risk } \rightarrow \text{ rates up } \rightarrow \text{ currency up}\\[2em]
%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
\qquad\quad\quad \underline{\mathbf{GROWTH\ INDICATORS}}\\[1em]
\qquad\quad\quad\quad \bullet \quad \text{GDP}\\
\qquad\quad\quad\quad \bullet \quad \text{PMI (Manufacturing \& Services) -- VERY important}\\
\qquad\quad\quad\quad \bullet \quad \text{Retail sales}\\[1em]
\qquad\quad\quad \text{Why PMI is important:}\\
\qquad\quad\quad\quad \bullet \quad \text{PMI > 50 } \rightarrow \text{ expansion}\\
\qquad\quad\quad\quad \bullet \quad \text{PMI < 50 } \rightarrow \text{ contraction}\\[1em]
\qquad\quad\quad \text{Logic: strong growth } \rightarrow \text{ tighter policy } \rightarrow \text{ currency up}\\[2em]
%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
\qquad\quad\quad \underline{\mathbf{CENTRAL\ BANK\ COMMUNICATION\ (CRITICAL)}}\\[1em]
\qquad\quad\quad \text{This is where expectations are formed:}\\[1em]
\qquad\quad\quad\quad \bullet \quad \text{Speeches (Powell, Lagarde)}\\
\qquad\quad\quad\quad \bullet \quad \text{Meeting minutes}\\
\qquad\quad\quad\quad \bullet \quad \text{Press conferences}\\[1em]
\qquad\quad\quad\quad \bullet \quad \text{Hawkish } \rightarrow \text{ rates up } \rightarrow \text{ currency up}\\
\qquad\quad\quad\quad \bullet \quad \text{Dovish } \rightarrow \text{ rates down } \rightarrow \text{ currency down}\\[1em]
\qquad\quad\quad \text{Note: sometimes words move markets more than data}\\[2em]
%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
\qquad\quad\quad \textbf{Market Expectations}\\[1em]
\qquad\quad\quad \text{The market already prices future interest rates.}\\[1em]
\qquad\quad\quad \textbf{Tools:}\\
\qquad\quad\quad\quad \bullet \quad \text{Interest rate futures}\\
\qquad\quad\quad\quad \bullet \quad \text{Bond yields (2Y, 10Y)}\\[1em]
\qquad\quad\quad \textbf{Key idea: Forex = pricing of future rate differentials}\\[2em]
%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
\qquad\quad\quad \textbf{Europe: Tracking expectations}\\[1em]

\qquad\quad\quad\quad \bullet \quad \text{Euribor futures}\\

\qquad\quad\quad\quad \bullet \quad \text{Directly linked to ECB expectations}\\[1em]
